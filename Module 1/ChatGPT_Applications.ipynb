## 1. Sử dung ChatGPT API

````python
# Tải và cập nhật thư viện openai bản mới nhất
!pip install --upgrade --quiet openai

# Import library
import openai
from openai import OpenAI
import IPython            # Dùng để xài tính năng hiển thị chữ thông qua markdown

# Set your OPENROUTER_API_KEY
OPENROUTER_API_KEY = "YOUR_API_KEY_HERE"

# Cấu hình OpenAI client để trỏ về OpenRouter
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",  # Dùng máy chủ của Open Router thay vì Open AI
    api_key= "YOUR_API_KEY_HERE",
)

def set_open_params(
    model="openai/gpt-oss-20b:free",  # Tên model AI lấy trên trang https://openrouter.ai/openai/gpt-oss-20b:free/api
    temperature=0.7,                  # Độ sáng tạo - range(0, 1)
    max_tokens=256,
    top_p=1,                          # Độ đa dạng trong từ vựng
    frequency_penalty=0,              # Phạt lặp từ
    presence_penalty=0,               # Cũng na ná phạt lặp từ
):
    """ set openrouter parameters"""

    openai_params = {}

    openai_params['model'] = model
    openai_params['temperature'] = temperature
    openai_params['max_tokens'] = max_tokens
    openai_params['top_p'] = top_p
    openai_params['frequency_penalty'] = frequency_penalty
    openai_params['presence_penalty'] = presence_penalty
    return openai_params

def get_completion(params, prompt):
    """ GET completion from openrouter api"""

    # client.chat.completions.create là lệnh gửi yêu cầu lên OpenRouter
    response = client.chat.completions.create(
        model=params['model'],
        messages=[{"role": "user", "content": prompt}], # Biến đổi prompt thành cấu trúc messages
        temperature=params['temperature'],
        max_tokens=params['max_tokens'],
        top_p=params['top_p'],
        frequency_penalty=params['frequency_penalty'],
        presence_penalty=params['presence_penalty'],
    )
    return response

def print_response(response):
    # Trích xuất dữ liệu từ cấu trúc JSON trả về mới của chuẩn Chat
    return IPython.display.Markdown(response.choices[0].message.content)

# Example:
# basic parameters
params = set_open_params()
prompt = "What is ChatGPT?"

response = get_completion(params, prompt)
print_response(response)
````

## 2. Sử dụng API ChatGPT thông qua thư viện `langchain`

````python
# Install library
!pip install --upgrade --quiet openai langchain langchain-openai langchain-core

# Import library
import os
import IPython
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    AIMessage,
    HumanMessage,
    SystemMessage
)

# 1. Khai báo API Key
OPENROUTER_API_KEY = "YOUR_API_KEY_HERE"

# 2. ÉP BUỘC LangChain nhận key thông qua biến môi trường (cách này chuẩn và ít lỗi nhất)
os.environ["OPENAI_API_KEY"] = OPENROUTER_API_KEY

# Cấu hình ChatOpenAI trỏ về OpenRouter (không cần truyền api_key vào đây nữa)
chat = ChatOpenAI(
    base_url="https://openrouter.ai/api/v1",
    model="openai/gpt-oss-20b:free",
    temperature=0
)

USER_INPUT = "I love AI VIET NAM."
FINAL_PROMPT = """Classify the text into neutral, negative or positive.
Text: {user_input}.
Sentiment:"""

# Gọi mô hình
response = chat.invoke([HumanMessage(content=FINAL_PROMPT.format(user_input=USER_INPUT))])

# In kết quả dạng nội dung text
print(response.content)
````
